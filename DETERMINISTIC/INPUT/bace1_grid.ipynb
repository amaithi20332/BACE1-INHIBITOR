```python
import os
import pandas as pd
dimension=pd.read_csv("dimensions_coordinates.csv")
file_name="config.txt"
ext=(".pdb")
check=[]
with open("configuration.txt","r") as fin:
    for files  in os.listdir("alpha_fold"):
        if files in dimension.values:
            pdb_name=files
            pdb=pdb_name.split("-")[1]
            directory=pdb
            parent_dir = "Vina"
            path = os.path.join(parent_dir, directory)
            os.mkdir(path)
            file_location=os.path.join(path,file_name)
            index= int(dimension[dimension['PDB_name']==pdb_name].index[0])
            X_c=str(dimension.X_center[index])
            Y_c=str(dimension.Y_center[index])
            Z_c=str(dimension.Z_center[index])
            X_d=str(dimension.X_dim[index])
            Y_d=str(dimension.Y_dim[index])
            Z_d=str(dimension.Z_dim[index])
            with open(file_location,"w") as fout:
                fout.write("receptor = "+pdb+".pdbqt"+"\n"+"\n"+"cpu=10"+"\n"+"seed =1"+"\n"+"num_modes=5"+"\n"+"exhaustiveness=5"+"\n"+"\n"+"\n"+"center_x = "+X_c+"\n"+"center_y ="+Y_c+"\n"+"center_z ="+Z_c+"\n"+"\n"+"size_x ="+X_d+"\n"+"size_y = "+Y_d+"\n"+"size_z = "+Z_d)      
        else:
            check.append(files)
```

```python
dict = {'double_coordinates': check}  
df = pd.DataFrame(dict) 
df.to_csv('Double_coordinates.csv')
```
